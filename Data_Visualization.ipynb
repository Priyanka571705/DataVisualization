```python
# first we need data to plot
x = [1,2,3,4,5]
y = [10,12,14,7,11]

plt.plot(x,y, label='Line plot', color='black', marker='s')
plt.title('Line plot')
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#BAR PLOT

categories = ['A','B','C','D']
values = [10,20,15,5]

plt.bar(categories, values, color= ['green', 'blue', 'red', 'yellow'], hatch='/')
plt.title('Bar Plot Example')
plt.xlabel('categories')
plt.ylabel('values')
plt.show()
```

```python
#SCATTER PLOT

study_hours = [1,2,3,4,5]
exam_scores = [55,66,80,90,34]

#scatter plot
plt.scatter(study_hours, exam_scores, color='black')
plt.title('tudy_hours, exam_score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')
plt.grid(True)
plt.show()
```

```python
fig, axs = plt.subplots(1,2, figsize=(10,4))

axs[0].bar(['A','B','C'],[10,20,15], hatch='.')
axs[0].set_title('Bar Plot')

axs[1].plot([1,2,3],[4,5,6])
axs[1].set_title('Line Plot')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
# sns.set
```

```python
df = sns.load_dataset('tips')  # build-in dataset
df.head()
```

```python
sns.countplot(data=df, x='day', hue='sex')
plt.title('count of Tips by Day and Gender')
plt.show()
```

```python
# create a box plot for the 'total_bill' column
sns.boxplot(y=df["total_bill"])
plt.title("Box plot of Total Bill")
plt.show()
```

```python
# create a sample list
data = [5,6,8,12,14,22,25,33,35,44,45,50,60]

# create a box plot
plt.boxplot(data)

# Add title and labels
plt.title("Box Plot Example")
plt.xlabel("Data")
plt.ylabel("Values")

plt.show()
```

```python
sns.boxplot(data=df, x='day', y='total_bill', hue="sex")
plt.title("Box Plot of Total Bill by Day")
plt.show()
```

```python
# Pie chart
name = ["apples", "Bananas", "cherries", "Data"]
sizes = [25,35,30,10]

plt.figure(figsize=(10,5))
plt.pie(sizes, labels=name, autopct="%1.1f%%", startangle=90, shadow=True)
plt.title("fruit Distribution")
plt.show()
```
